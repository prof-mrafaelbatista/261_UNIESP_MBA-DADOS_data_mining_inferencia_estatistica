## Importações de Bibliotecas

```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/prof-mrafaelbatista/261_UNIESP_MBA-DADOS_data_mining_inferencia_estatistica/refs/heads/main/datasets/OnlineRetail.csv"
```

```python
df = pd.read_csv(url, encoding='latin1')
```

```python
# Amostragem simples
amostra_simples = df.sample(n=1000, random_state=42)
print(f"Amostra: {amostra_simples.shape}")
```

```python
# Amostragem estratificada por país (top 5)
top5 = df['Country'].value_counts().head(5).index
df_top5 = df[df['Country'].isin(top5)]
amostra_estrat = df_top5.groupby('Country', group_keys=False).apply(
    lambda x: x.sample(frac=0.05, random_state=42)
)
```

```python
print("Amostra estratificada por país:")
print(amostra_estrat['Country'].value_counts())
```

```python
import csv
import requests
from deep_translator import GoogleTranslator
```

```python
URL_SLIP = 'https://api.adviceslip.com/advice'
```

```python
lista_de_conselhos = []
```

```python
for i in range(10):
    resp = requests.get(URL_SLIP)
    conselho = resp.json()['slip']['advice']
    lista_de_conselhos.append(conselho)
    
```

```python
print(lista_de_conselhos)
```

```python
# print(resp.json())
# print(resp.json()['slip'])
# print(resp.json()['slip']['advice'])
```

```python
lista_traduzida = []
```

```python
for conselho in lista_de_conselhos:
    traducao = GoogleTranslator(
        source='english',
        target='portuguese').translate(conselho)
    lista_traduzida.append(traducao)

print(lista_traduzida)
```

```python
with open('conselhos.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['conselho'])
    for conselho in lista_traduzida:
        writer.writerow([conselho])

print("Arquivo salvo: conselhos.csv")
```
